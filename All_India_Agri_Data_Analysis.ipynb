```python
# Check reported yield against production divided by area

for year in ["2017 - 2018", "2018 - 2019", "2019 - 2020",
             "2020 - 2021", "2021 - 2022", "2022 - 2023"]:

    area_col = next(
        col for col in df.columns
        if col.startswith(year) and "Area" in col
    )

    production_col = next(
        col for col in df.columns
        if col.startswith(year) and "Production" in col
    )

    yield_col = next(
        col for col in df.columns
        if col.startswith(year) and "Yield" in col
    )

    valid = (
        df[area_col].notna()
        & df[production_col].notna()
        & df[yield_col].notna()
        & (df[area_col] > 0)
    )

    calculated_yield = df.loc[valid, production_col] / df.loc[valid, area_col]

    difference = (calculated_yield - df.loc[valid, yield_col]).abs()

    print(
        year,
        "| Records checked:", valid.sum(),
        "| Records with difference > 0.1:", (difference > 0.1).sum()
    )
```

```python
# Find records where reported and calculated yields differ by more than 0.1

for year in ["2017 - 2018", "2018 - 2019", "2019 - 2020",
             "2020 - 2021", "2021 - 2022", "2022 - 2023"]:

    area_col = next(c for c in df.columns
                    if c.startswith(year) and "Area" in c)
    production_col = next(c for c in df.columns
                          if c.startswith(year) and "Production" in c)
    yield_col = next(c for c in df.columns
                     if c.startswith(year) and "Yield" in c)

    valid = (
        df[area_col].notna()
        & df[production_col].notna()
        & df[yield_col].notna()
        & (df[area_col] > 0)
    )

    calculated = df.loc[valid, production_col] / df.loc[valid, area_col]
    difference = (calculated - df.loc[valid, yield_col]).abs()

    issues = df.loc[valid].loc[difference > 0.1,
                               ["State", "District", area_col,
                                production_col, yield_col]].copy()

    if not issues.empty:
        issues["Calculated Yield"] = calculated[difference > 0.1]
        issues["Year"] = year
        print(f"\nYear: {year}")
        print(issues.to_string(index=False))
```

```python
# Create a cleaned copy of the dataset
cleaned_df = df.copy()

# Remove the temporary column used for missing-value analysis
if "Missing_Count" in cleaned_df.columns:
    cleaned_df = cleaned_df.drop(columns=["Missing_Count"])

# Confirm the final dataset
print("Original records:", len(df))
print("Cleaned records:", len(cleaned_df))
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Missing values retained:", cleaned_df.isnull().sum().sum())

# Save the cleaned dataset
cleaned_df.to_csv("agricultural_data_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

```python

import pandas as pd

# Column names for 2022–2023
area_col = '2022 - 2023 - Area( Ha. )'
production_col = '2022 - 2023 - Production  (Tonnes)'

# Check that the required columns exist
print("Checking columns...")
print(area_col in df.columns)
print(production_col in df.columns)

# Prepare state-wise data
state_data = df[
    ['State', area_col, production_col]
].dropna(subset=['State', area_col, production_col])

# Calculate state-wise totals
state_summary = state_data.groupby('State').agg(
    Total_Area_Ha=(area_col, 'sum'),
    Total_Production_Tonnes=(production_col, 'sum'),
    District_Records=('State', 'count')
).reset_index()

# Calculate weighted yield
state_summary['Weighted_Yield'] = (
    state_summary['Total_Production_Tonnes'] /
    state_summary['Total_Area_Ha']
)

# Select top 10 states
top_10_states = state_summary.sort_values(
    by='Total_Production_Tonnes',
    ascending=False
).head(10)

display(top_10_states.round(2))
```

```python

import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Sort for a horizontal bar chart
chart_data = top_10_states.sort_values(
    'Total_Production_Tonnes',
    ascending=True
).copy()

plt.figure(figsize=(11, 7))

plt.barh(
    chart_data['State'],
    chart_data['Total_Production_Tonnes'] / 1e9
)

plt.title('Top 10 States by Recorded Agricultural Production (2022–23)')
plt.xlabel('Production (Billion Tonnes)')
plt.ylabel('State')
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python

# Inspect original 2022–23 records for the top 3 states

top_3_names = top_10_states.head(3)['State'].tolist()

inspection = df[
    df['State'].isin(top_3_names)
][[
    'State',
    'District',
    area_col,
    production_col
]].dropna(
    subset=[area_col, production_col]
)

display(
    inspection.sort_values(
        production_col,
        ascending=False
    ).head(15)
)
```

```python

# Find all yield-related columns
yield_columns = [
    col for col in df.columns
    if 'Yield' in col or 'yield' in col
]

print("Yield columns found:")
for col in yield_columns:
    print(col)

# Inspect the latest year's original yield values
display(
    df[['State', 'District'] + yield_columns[-1:]].head(15)
)
```

```python

# Top 10 districts by recorded production in 2022–23

district_summary = df[
    ['State', 'District', area_col, production_col]
].dropna(
    subset=['State', 'District', area_col, production_col]
).copy()

district_summary['Calculated_Yield'] = (
    district_summary[production_col] /
    district_summary[area_col]
)

top_10_districts = district_summary.sort_values(
    by=production_col,
    ascending=False
).head(10)

display(top_10_districts.round(2))
```

```python

plt.figure(figsize=(12, 7))

plt.barh(
    top_10_districts['District'] + ', ' + top_10_districts['State'],
    top_10_districts[production_col] / 1e9
)

plt.title('Top 10 Districts by Recorded Production (2022–23)')
plt.xlabel('Recorded Production (Billions of Source Units)')
plt.ylabel('District')
plt.gca().invert_yaxis()
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python

# Compare reported yield with production divided by area

yield_col = '2022 - 2023 - Yield  (Ton./Ha.)'

yield_check = df[
    ['State', 'District', area_col, production_col, yield_col]
].dropna(
    subset=[area_col, production_col, yield_col]
).copy()

yield_check['Calculated_Ratio'] = (
    yield_check[production_col] / yield_check[area_col]
)

yield_check['Difference'] = (
    yield_check['Calculated_Ratio'] - yield_check[yield_col]
)

yield_check['Ratio_to_Reported'] = (
    yield_check['Calculated_Ratio'] / yield_check[yield_col]
)

display(
    yield_check[
        ['State', 'District', area_col, production_col,
         yield_col, 'Calculated_Ratio', 'Difference',
         'Ratio_to_Reported']
    ].head(20).round(3)
)

print("Reported yield statistics:")
display(yield_check[yield_col].describe().round(2))

print("Calculated production/area statistics:")
display(yield_check['Calculated_Ratio'].describe().round(2))
```

```python

yield_col = '2022 - 2023 - Yield  (Ton./Ha.)'

# Match the top 10 districts with their original reported yields
top_10_yield_check = df[
    ['State', 'District', area_col, production_col, yield_col]
].merge(
    top_10_districts[['State', 'District']],
    on=['State', 'District'],
    how='inner'
)

top_10_yield_check['Calculated_Ratio'] = (
    top_10_yield_check[production_col] /
    top_10_yield_check[area_col]
)

top_10_yield_check['Difference'] = (
    top_10_yield_check['Calculated_Ratio'] -
    top_10_yield_check[yield_col]
)

display(
    top_10_yield_check[
        ['State', 'District', yield_col,
         'Calculated_Ratio', 'Difference']
    ].round(2)
)
```

```python

# Investigate the 15 highest reported yield values

yield_col = '2022 - 2023 - Yield  (Ton./Ha.)'

high_yield = df[
    ['State', 'District', area_col, production_col, yield_col]
].dropna(subset=[yield_col]).copy()

top_yield_districts = high_yield.sort_values(
    by=yield_col,
    ascending=False
).head(15)

display(top_yield_districts.round(2))

# Calculate IQR-based outlier threshold
Q1 = high_yield[yield_col].quantile(0.25)
Q3 = high_yield[yield_col].quantile(0.75)
IQR = Q3 - Q1

upper_limit = Q3 + (1.5 * IQR)

outliers = high_yield[high_yield[yield_col] > upper_limit]

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Upper outlier limit:", round(upper_limit, 2))
print("Number of high-yield outliers:", len(outliers))
```

```python

plt.figure(figsize=(10, 5))

plt.hist(
    high_yield[yield_col],
    bins=40,
    edgecolor='black'
)

plt.title('Distribution of Agricultural Yield (2022–23)')
plt.xlabel('Yield (Tonnes per Hectare, as labelled in source)')
plt.ylabel('Number of District Records')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Summarise high-yield records by state

outlier_state_summary = outliers.groupby('State').agg(
    Outlier_Records=(yield_col, 'count'),
    Average_Yield=(yield_col, 'mean'),
    Maximum_Yield=(yield_col, 'max')
).reset_index()

outlier_state_summary = outlier_state_summary.sort_values(
    by='Outlier_Records',
    ascending=False
)

display(outlier_state_summary.round(2))

print("Total flagged records:", len(outliers))
print("States with flagged records:", outliers['State'].nunique())
```

```python

plt.figure(figsize=(10, 5))

plt.hist(
    high_yield[yield_col],
    bins=40,
    edgecolor='black'
)

plt.xscale('log')
plt.title('Agricultural Yield Distribution – Log Scale (2022–23)')
plt.xlabel('Yield (Tonnes per Hectare, as labelled in source)')
plt.ylabel('Number of District Records')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Compare flagged records with total valid records in each state

total_by_state = high_yield.groupby('State').size().reset_index(
    name='Total_Valid_Records'
)

outliers_by_state = outliers.groupby('State').size().reset_index(
    name='Outlier_Records'
)

state_outlier_rate = total_by_state.merge(
    outliers_by_state,
    on='State',
    how='left'
)

state_outlier_rate['Outlier_Records'] = (
    state_outlier_rate['Outlier_Records'].fillna(0).astype(int)
)

state_outlier_rate['Outlier_Percentage'] = (
    state_outlier_rate['Outlier_Records'] /
    state_outlier_rate['Total_Valid_Records'] * 100
)

state_outlier_rate = state_outlier_rate.sort_values(
    'Outlier_Percentage',
    ascending=False
)

display(state_outlier_rate.round(2))
```

```python

plt.figure(figsize=(11, 7))

plt.barh(
    state_outlier_rate['State'],
    state_outlier_rate['Outlier_Percentage']
)

plt.title('Percentage of High-Yield Outlier Records by State (2022–23)')
plt.xlabel('Outlier Records (%)')
plt.ylabel('State')
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Filter states with at least 10 valid records
reliable_state_comparison = state_outlier_rate[
    state_outlier_rate['Total_Valid_Records'] >= 10
].sort_values(
    'Outlier_Percentage',
    ascending=True
)

plt.figure(figsize=(11, 7))

plt.barh(
    reliable_state_comparison['State'],
    reliable_state_comparison['Outlier_Percentage']
)

plt.title('State-wise High-Yield Outlier Percentage (2022–23)')
plt.xlabel('Outlier Records (%)')
plt.ylabel('State')
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
